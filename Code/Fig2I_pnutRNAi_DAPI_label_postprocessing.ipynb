# Fig. 2 pnut(RNAi) label post-processing

Cleaned from `RingCanals/Code/Pnut_LabelProcessing.ipynb`. This notebook loads labels generated or corrected in napari and saves finalized DAPI label stacks for Fig. 2I ploidy quantification.


## Imports

Figure association: final label preparation for Fig. 2I.


```python
from __future__ import annotations

from pathlib import Path

import napari
from csbdeep.io import save_tiff_imagej_compatible
from tifffile import imread

%matplotlib inline
```

## Paths


```python
# Fig. 2I processed image/label folder generated by napari correction.
data_directory = Path("../Images/Pnut_phenotype/Control/control_same_exposure/ProcessedImages/AndersImages/control final")
results_directory = Path("../Images/Pnut_phenotype/Results")
expt_name = "Pnut_control"

img_files = sorted(data_directory.glob("*img.tif"))
label_files = sorted(data_directory.glob("*labels.tif"))
img_list = [imread(path) for path in img_files]
label_list = [imread(path) for path in label_files]

print(f"Loaded {len(img_list)} images and {len(label_list)} label stacks")
```

## Optional napari review


```python
# Fig. 2I QC: inspect one image/label pair before saving finalized labels.
position = 2
viewer = napari.Viewer()
viewer.add_image(img_list[position], name="Image", channel_axis=-1)
viewer.add_labels(label_list[position], name="Label")
napari.run()
```

## Save finalized labels


```python
# Fig. 2I output: write each finalized label image next to its source label stack.
for label_path, labels in zip(label_files, label_list):
    original_stem = Path(label_path).stem
    save_path = data_directory / f"{original_stem}_p.tif"
    save_tiff_imagej_compatible(save_path, labels, axes="ZYX")
    print(f"Saved {save_path}")
```
